```python
import pandas as pd
import tabulate
```

## Functions

```python
def create_report(midx):
    swin_df = pd.read_csv(f"SwinUnet-{midx}.csv")
    unet_df = pd.read_csv(f"UNet-{midx}.csv")
    deeplab_df = pd.read_csv(f"DeepLab-{midx}.csv")

    print(tabulate.tabulate([
            ["Swin Transformer", swin_df["train_loss"].min(), 
             swin_df["val_loss"].min(), swin_df["train_acc"].max(), 
             swin_df["val_acc"].max(), swin_df["train_iou"].max(),
             swin_df["val_iou"].max()],
            ["Unet", unet_df["train_loss"].min(), 
             unet_df["val_loss"].min(), unet_df["train_acc"].max(), 
             unet_df["val_acc"].max(), unet_df["train_iou"].max(),
             unet_df["val_iou"].max()],
            ["DeepLab", deeplab_df["train_loss"].min(), 
             deeplab_df["val_loss"].min(), deeplab_df["train_acc"].max(), 
             deeplab_df["val_acc"].max(), deeplab_df["train_iou"].max(),
             deeplab_df["val_iou"].max()],
        ],
        headers=["Model", "Train Loss", "Val Loss", "Train Acc", "Val Acc", "Train IoU", "Val IoU"]))
```

## Lr=1e-3

```python
create_report(1)
```

## Lr=1e-4

```python
create_report(2)
```

## Lr=1e-5

```python
create_report(3)
```
